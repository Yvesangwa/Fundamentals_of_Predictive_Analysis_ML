## **1. Data Format Conversion**

### **Create a dataset in CSV format**

```python
# Import libraries

import pandas as pd
import random
```

```python
# Craete the lists

names = [
    "Amina", "Kevin", "Sofia", "Patrick", "Nadia",
    "Samuel", "Chloe", "Eric", "Maya", "Lucas"]

cities = [
    "Ottawa", "Toronto", "Montreal", "Vancouver", "Calgary",
    "Edmonton", "Winnipeg", "Halifax", "Quebec City", "Victoria"]
```

```python
# Generate 100 users

fictional_user_data = []

for i in range(100):
    name = random.choice(names)
    age = random.randint(18, 70)
    city = random.choice(cities)
    
    fictional_user_data.append([name, age, city])
```

```python
# Create a dataframe

fictional_user_df = pd.DataFrame(fictional_user_data, columns=["Name", "Age", "City"])

fictional_user_df.head()
```

```python
# Save the dataset as csv format

fictional_user_df.to_csv("fictional_user_df.csv", index=False)
```

### **Convert the CSV file to Parquet format**

```python
fictional_user_df.to_parquet("fictional_user_df.parquet", index=False)
```

### **Load the parquet file back**

```python
df_parquet = pd.read_parquet("fictional_user_df.parquet")

df_parquet.head()
```

## **2. Implementing ETL Process**

#### **Extract data from csv file**

```python
# Extract data from csv file

fictional_user_data = pd.read_csv("fictional_user_df.csv")
```

```python
fictional_user_data.head()
```

#### **Transform the data by filtering out records where the age is below 18**

```python
# Transform the data by filtering out records where the age is below 18 

fictional_user_filtered = fictional_user_data[fictional_user_data["Age"] >= 18]
```

```python
fictional_user_filtered.tail()
```

```python
fictional_user_filtered["Age"].min()
```

#### **Converting the names to uppercase**

```python
fictional_user_filtered["Name"] = fictional_user_filtered["Name"].str.upper()
```

```python
fictional_user_filtered.head()
```

#### **Load and save the transformed data into a new CSV file**

```python
fictional_user_filtered.to_csv("transformed_data.csv", index=False)
```

```python
# Verify the transformation by loading and displaying the new file.

df_transformed = pd.read_csv("transformed_data.csv")

df_transformed.head()
```

## **3.Exploring Data Models with SQLite and MongoDB**

#### **Install and connect to a SQLite database**

```python
import sqlite3
```

```python
# Create a new database

conn = sqlite3.connect("user_database.db")

cursor = conn.cursor()
```

#### **Create the relational table**

```python
# Create the relational table

cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Name TEXT NOT NULL,
    Age INTEGER,
    City TEXT)
""")

conn.commit()
```

#### **Insert 5 records**

```python
users_data = [
    ("Alice", 25, "Ottawa"),
    ("David", 32, "Toronto"),
    ("Amina", 28, "Montreal"),
    ("Samuel", 41, "Calgary"),
    ("Sofia", 35, "Vancouver")
]

cursor.executemany(
    "INSERT INTO users (Name, Age, City) VALUES (?, ?, ?)",
    users_data
)

conn.commit()
```

#### **Retrieve the data using an SQL query**

```python
cursor.execute("SELECT * FROM users")

rows = cursor.fetchall()

for row in rows:
    print(row)
```
