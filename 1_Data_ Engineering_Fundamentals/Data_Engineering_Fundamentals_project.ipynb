## **1. Data Format Conversion**

### **Create a dataset in CSV format**

```python
# Import libraries

import pandas as pd
import random
```

```python
# Craete the lists

names = [
    "Amina", "Kevin", "Sofia", "Patrick", "Nadia",
    "Samuel", "Chloe", "Eric", "Maya", "Lucas"]

cities = [
    "Ottawa", "Toronto", "Montreal", "Vancouver", "Calgary",
    "Edmonton", "Winnipeg", "Halifax", "Quebec City", "Victoria"]
```

```python
# Generate 100 users

fictional_user_data = []

for i in range(100):
    name = random.choice(names)
    age = random.randint(18, 70)
    city = random.choice(cities)
    
    fictional_user_data.append([name, age, city])
```

```python
# Create a dataframe

fictional_user_df = pd.DataFrame(fictional_user_data, columns=["Name", "Age", "City"])

fictional_user_df.head()
```

```python
# Save the dataset as csv format

fictional_user_df.to_csv("fictional_user_df.csv", index=False)
```

### **Convert the CSV file to Parquet format**

```python
fictional_user_df.to_parquet("fictional_user_df.parquet", index=False)
```

### **Load the parquet file back**

```python
df_parquet = pd.read_parquet("fictional_user_df.parquet")

df_parquet.head()
```

## **2. Implementing ETL Process**

#### **Extract data from csv file**

```python
# Extract data from csv file

fictional_user_data = pd.read_csv("fictional_user_df.csv")
```

```python
fictional_user_data.head()
```

#### **Transform the data by filtering out records where the age is below 18**

```python
# Transform the data by filtering out records where the age is below 18 

fictional_user_filtered = fictional_user_data[fictional_user_data["Age"] >= 18]
```

```python
fictional_user_filtered.tail()
```

```python
fictional_user_filtered["Age"].min()
```

#### **Converting the names to uppercase**

```python
fictional_user_filtered["Name"] = fictional_user_filtered["Name"].str.upper()
```

```python
fictional_user_filtered.head()
```

#### **Load the transformed data into a new CSV file**

```python
filtered_data = fictional_user_filtered.to_csv("filtered_data.csv", index=False)
```
